```python
import pandas as pd
import matplotlib.pyplot as plt
```

Loss every epoch

```python
df_e = pd.read_csv('epoch_metrics.csv')

plt.figure(figsize=(8, 5))

plt.plot(df_e['epoch'], df_e['train_loss'], marker='o', label='Train')
plt.plot(df_e['epoch'], df_e['val_loss'], marker='o', label='Validation')

plt.xlabel('epoch')
plt.ylabel('loss')
plt.title('Train vs Validation Loss')
plt.legend()
plt.grid(True)

plt.show()
```
